#### Compare .npy files

```python
import os
import numpy as np


def compare_two_npy(file1, file2, rtol=1e-5, atol=1e-8):
    """
    Compare the content of two .npy files using NumPy built-in functions.
    Returns one of: "exact", "close", "different".
    """
    a = np.load(file1, allow_pickle=False)
    b = np.load(file2, allow_pickle=False)

    if np.array_equal(a, b, equal_nan=True):
        print(f"[EXACT] {os.path.basename(file1)} == {os.path.basename(file2)}")
        return "exact"
    elif np.allclose(a, b, rtol=rtol, atol=atol, equal_nan=True):
        print(f"[CLOSE] {os.path.basename(file1)} ≈ {os.path.basename(file2)} (within tolerance)")
        return "close"
    else:
        print(f"[DIFFER] {os.path.basename(file1)} != {os.path.basename(file2)}")
        return "different"


def compare_npy_folders(folder1, folder2, rtol=1e-5, atol=1e-8, recursive=False):
    """
    Compare all .npy files in two folders using NumPy built-in functions.

    Parameters
    ----------
    folder1 : str
        Path to the first folder.
    folder2 : str
        Path to the second folder.
    rtol : float, optional
        Relative tolerance for np.allclose, by default 1e-5.
    atol : float, optional
        Absolute tolerance for np.allclose, by default 1e-8.
    recursive : bool, optional
        If True, compare subdirectories by relative paths.

    Returns
    -------
    dict
        A summary dictionary with counts for each comparison result.
    """
    if recursive:
        npy_files = set()
        for root, _, files in os.walk(folder1):
            for f in files:
                if f.endswith(".npy"):
                    rel = os.path.relpath(os.path.join(root, f), folder1)
                    npy_files.add(rel)
        for root, _, files in os.walk(folder2):
            for f in files:
                if f.endswith(".npy"):
                    rel = os.path.relpath(os.path.join(root, f), folder2)
                    npy_files.add(rel)
    else:
        npy_files = set(f for f in os.listdir(folder1) if f.endswith(".npy"))
        npy_files |= set(f for f in os.listdir(folder2) if f.endswith(".npy"))

    results = {"exact": 0, "close": 0, "different": 0, "missing": 0}
    print(f"Comparing {len(npy_files)} files...")

    for f in sorted(npy_files):
        path1 = os.path.join(folder1, f)
        path2 = os.path.join(folder2, f)

        if not os.path.exists(path1) or not os.path.exists(path2):
            print(f"[MISSING] {f}")
            results["missing"] += 1
            continue

        result = compare_two_npy(path1, path2, rtol=rtol, atol=atol)
        results[result] += 1

    print("\nSummary:")
    for key, value in results.items():
        print(f"{key:<10}: {value}")
    return results
```
